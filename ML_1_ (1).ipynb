# Outlier problem

*   List item
*   List item


# IQR (Interquartile Range) = Q3 - Q1
# Lower outlier threshold: Q1 - 1.5 * IQR
# Upper outlier threshold: Q3 + 1.5 * IQR


```python
#An outlier is a data point that is very different from the other values in a dataset.
#An outlier is an unusually high or unusually low value compared with the rest of the data.
```

```python
df_tip = df_1["tip"]
df_tip.describe().T
```

```python
Q1 = df_tip.quantile(0.25)
Q3 = df_tip.quantile(0.75)
IQR = Q3-Q1
```

```python
below = Q1- 1.5*IQR
above = Q3 + 1.5*IQR
```

```python
below
```

```python
df_1[df_1["tip"] > above]
```

```python
# Save clear data
clear_df = df_1[df_1["tip"] < above].dropna().reset_index()
```
