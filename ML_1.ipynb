## 6. Outliers: the IQR method
An **outlier** is an unusually high or low value compared with the rest of the data.

- **IQR** (Interquartile Range) = Q3 − Q1
- **Lower fence** = Q1 − 1.5 × IQR
- **Upper fence** = Q3 + 1.5 × IQR

Values outside the fences are outliers.

```python
df_tip = df_1["tip"]
df_tip.describe()
```

```python
# Quartiles, IQR and fences
Q1 = df_tip.quantile(0.25)
Q3 = df_tip.quantile(0.75)
IQR = Q3 - Q1

below = Q1 - 1.5 * IQR
above = Q3 + 1.5 * IQR
below, above
```

```python
# Rows with outlier tips
df_1[df_1["tip"] > above]
```

```python
# Keep only values inside the fences and renumber the rows
clear_df = df_1[df_1["tip"].between(below, above)].reset_index(drop=True)
clear_df['tip'].describe()
```
